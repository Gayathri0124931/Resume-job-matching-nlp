```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
import nltk

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    if pd.isna(text):
        return ""
    
    # Convert to lowercase
    text = text.lower()
    
    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)
    
    # Remove special characters and numbers
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()
    
    # Tokenization
    tokens = nltk.word_tokenize(text)
    
    # Stop-word removal and lemmatization
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
        if word not in stop_words
    ]
    
    return " ".join(tokens)
```

```python
sample_resume = cv_df["resume_text"].iloc[0]

print("===== ORIGINAL RESUME =====")
print(sample_resume[:1000])

print("\n===== PREPROCESSED RESUME =====")
print(preprocess_text(sample_resume)[:1000])
```

```python
cv_df["processed_resume"] = cv_df["resume_text"].apply(preprocess_text)

job_df["processed_job"] = job_df["job_description"].apply(preprocess_text)

print("Resume preprocessing completed.")
print("Job description preprocessing completed.")
```

```python
import ast

def extract_skills(skill_text):
    try:
        return ast.literal_eval(skill_text)
    except:
        return []

cv_df["skills_list"] = cv_df["clean_skills"].apply(extract_skills)
job_df["skills_list"] = job_df["clean_skills"].apply(extract_skills)
```

```python
from collections import Counter

candidate_skills = Counter()

for skills in cv_df["skills_list"]:
    candidate_skills.update(skills)

print("===== TOP 20 CANDIDATE SKILLS =====")

for skill, count in candidate_skills.most_common(20):
    print(skill, ":", count)
```

```python
job_skills = Counter()

for skills in job_df["skills_list"]:
    job_skills.update(skills)

print("===== TOP 20 REQUIRED JOB SKILLS =====")

for skill, count in job_skills.most_common(20):
    print(skill, ":", count)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2)
)

resume_tfidf = tfidf.fit_transform(
    cv_df["processed_resume"].fillna("")
)

print("TF-IDF Resume Matrix Shape:", resume_tfidf.shape)
```

```python
job_tfidf = tfidf.transform(
    job_df["processed_job"].fillna("")
)

print("TF-IDF Job Matrix Shape:", job_tfidf.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Compare first resume with first job
similarity = cosine_similarity(
    resume_tfidf[0],
    job_tfidf[0]
)

print("Resume-Job Similarity:", similarity[0][0])
```

```python
print("Existing matching information:")
display(matches_df.head())
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Select one resume
resume_index = 0

# Calculate similarity between the selected resume and all jobs
similarity_scores = cosine_similarity(
    resume_tfidf[resume_index],
    job_tfidf
).flatten()

print("Number of similarity scores:", len(similarity_scores))
```

```python
ranking_df = pd.DataFrame({
    "job_id": job_df["job_id"],
    "job_title": job_df["job_title"],
    "category": job_df["category"],
    "similarity_score": similarity_scores
})

# Sort from highest to lowest similarity
ranking_df = ranking_df.sort_values(
    by="similarity_score",
    ascending=False
).reset_index(drop=True)

# Add rank
ranking_df.insert(0, "rank", ranking_df.index + 1)

display(ranking_df.head(10))
```

```python
print("===== TOP 10 JOBS FOR SELECTED RESUME =====")

display(
    ranking_df[
        ["rank", "job_id", "job_title", "category", "similarity_score"]
    ].head(10)
)
```

```python
top_job_id = ranking_df.loc[0, "job_id"]

top_job = job_df[job_df["job_id"] == top_job_id].iloc[0]

print("===== BEST MATCH =====")
print("Job Title:", top_job["job_title"])
print("Category:", top_job["category"])
print("Similarity Score:", ranking_df.loc[0, "similarity_score"])

print("\nJob Description:")
print(top_job["job_description"])
```

```python
def rank_jobs_for_resume(resume_index, top_n=10):
    
    scores = cosine_similarity(
        resume_tfidf[resume_index],
        job_tfidf
    ).flatten()
    
    result = pd.DataFrame({
        "job_id": job_df["job_id"],
        "job_title": job_df["job_title"],
        "category": job_df["category"],
        "similarity_score": scores
    })
    
    result = result.sort_values(
        "similarity_score",
        ascending=False
    ).reset_index(drop=True)
    
    result.insert(0, "rank", result.index + 1)
    
    return result.head(top_n)
```

```python
top_jobs = rank_jobs_for_resume(0, 10)

display(top_jobs)
```
